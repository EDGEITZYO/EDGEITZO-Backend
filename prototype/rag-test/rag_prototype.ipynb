```python
# RAG 검색 테스트 - 질의랑 가장 유사한 논문 2개 순서대로 출력하기
results = collection.query(
    query_texts=["생명공학 교육과정 분석"],
    n_results=2
)

for i, (doc, meta) in enumerate(zip(results["documents"][0], results["metadatas"][0])):
    print(f"[{i+1}위]")
    print(f"제목: {meta['title']}")
    print(f"초록 앞부분: {doc[:100]}")
    print()
```

```python
test_queries = [
    "유전자 연구 윤리",
    "언론의 과학 보도",
    "대학생 생명공학 인식"
]

for query in test_queries:
    print(f"질의: {query}")
    results = collection.query(
        query_texts=[query],
        n_results=2
    )
    for i, meta in enumerate(results["metadatas"][0]):
        print(f"  [{i+1}위] {meta['title']}")
    print()
```
